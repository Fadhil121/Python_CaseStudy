```python
students = np.array(["Anu", "Binu", "Cathy",
                     "Deepa", "Ebin", "Fahad"])

marks = np.array([
    [82, 75, 91, 88],
    [61, 68, 70, 65],
    [45, 52, 48, 55],
    [90, 94, 89, 92],
    [72, 80, 76, 85],
    [38, 42, 49, 44]
])

subjects = np.array(["Python", "SQL",
                     "Excel", "Statistics"])
```

```python
# Total and average marks for each student
total = np.sum(marks, axis=1)
average = np.mean(marks, axis=1)
print("Total Marks:", total)
print("Average Marks:", average)
```

```python
# Subject averages
subject_avg = np.mean(marks, axis=0)
print("Subject Averages:", subject_avg)
```

```python
# Highest mark in each subject
highest_marks = np.max(marks, axis=0)
print("Highest Marks:", highest_marks)
```

```python
# Pass/fail status for every subject
status = np.where(marks >= 50, "Pass", "Fail")
print(status)
```

```python
# Students who passed every subject
fullpass = np.all(marks >= 50, axis=1)
print("Passed All Subjects:", students[fullpass])
```

```python
# Students whose average is below 60
below_60 = students[average < 60]
print("Average Below 60:", below_60)
```

```python
employees = np.array([
    "Arun", "Bhavana", "Charan",
    "Divya", "Eshan", "Fathima"
])

performance = np.array([
    [78, 82, 85, 88],
    [65, 70, 68, 74],
    [92, 89, 95, 91],
    [55, 62, 60, 67],
    [81, 79, 84, 87],
    [72, 75, 78, 80]
])

quarters = np.array(["Q1", "Q2", "Q3", "Q4"])
```

```python
# Total performance score of every employee
total = np.sum(performance, axis=1)
print("Total Scores:", total)
```

```python
# Average quarterly performance per employee
avg = np.mean(performance, axis=1)
print("Employee Averages:", avg)
```

```python
# Average performance for each quarter
quarter_avg = np.mean(performance, axis=0)
print("Quarterly Averages:", quarter_avg)
```

```python
# Employee with highest total score
highest = np.argmax(total)
print("Highest Performer:", employees[highest])
```

```python
# Rank employees by total score
ranking = np.argsort(total)[::-1]
print("Ranking:", employees[ranking])
```

```python
# Quarter-over-quarter percentage growth
growth = (np.diff(performance, axis=1) / performance[:, :-1]) * 100
print("Percentage Growth:")
print(np.round(growth, 2))
```

```python
# Employees with average >= 80
print("Average >= 80:", employees[avg >= 80])
```

```python
# Count employees with average below 70
count = np.sum(avg < 70)
print("Count Below 70:", count)
```

```python
# Highest score in entire dataset
maximum = np.max(performance)
print("Highest Score:", maximum)
```

```python
# Employee and quarter of highest score
row, col = np.where(performance == np.max(performance))

print("Employee:", employees[row])
print("Quarter:", quarters[col])
```
